```python
import re

def mask(string, pattern, mask_format):
    return re.sub(pattern, mask_format, string)
```

```python
# Example string and pattern
string = "My phone number is 123-456-7890"
pattern = r"\d{3}-\d{3}-\d{4}"

# Masking the phone number
masked_string = mask(string, pattern, "(###) ###-####")

print(masked_string)
```

```python
import masking.maskstring as msk

string = "produto para uso telefone: (39) 76877690 nome: Joao código: 7858769 modelo tal de tal"

# Anonimizing the phone number
phone_number_anonymized = msk.maskstring(string, "(###) ###-####", 0, 14)

# Anonimizing the name
name_anonymized = msk.maskstring(string, "###", 23, 27)

# Anonimizing the code
code_anonymized = msk.maskstring(string, "###", 38, 41)

# Combining the anonymized parts
final_string = phone_number_anonymized + " nome: " + name_anonymized + " código: " + code_anonymized + " modelo tal de tal"

print(final_string)
```

```python
import re

string = "produto para uso telefone: (39) 76877690 nome: Silva Joao crea: SP-1-123456789-A modelo tal de tal que usa tal"

def mask_sensitive_data(text):
    name_pattern = re.compile(r'\bnome:\s*(\w+\s+\w+)', re.IGNORECASE)
    crea_pattern = re.compile(r'\bcrea:\s*([A-Z0-9-]+)', re.IGNORECASE)
    phone_pattern = re.compile(r'\btelefone:\s*\(\d+\)\s*(\d+)', re.IGNORECASE)
    masked_text = re.sub(name_pattern, 'nome: [MASKED]', text)
    masked_text = re.sub(crea_pattern, 'crea: [MASKED]', masked_text)
    masked_text = re.sub(phone_pattern, 'telefone: [MASKED]', masked_text)

    return masked_text

masked_result = mask_sensitive_data(string)
print(masked_result)
```

```python
import spacy

string = "produto para uso telefone: (39) 76877690 nome: Silva Joao crea: SP-1-123456789-A modelo tal de tal que usa tal"

def mask_sensitive_data(text):
    nlp = spacy.load('pt_core_news_sm')
    doc = nlp(text)

    masked_text = text
    for ent in doc.ents:
        if ent.label_ == 'PER':
            masked_text = masked_text.replace(ent.text, "[MASKED]")
        elif ent.label_ == "PHONE":
            masked_text = masked_text.replace(ent.text, "[MASKED]")
        elif ent.text.lower() == 'crea':
            masked_text = masked_text.replace(ent.text, "[MASKED]")

    return masked_text


masked_result = mask_sensitive_data(string)
print(masked_result)
```
